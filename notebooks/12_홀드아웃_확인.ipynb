## 2. 🚨 holdout 을 엽니다

```python
df = labels.load(env.work_root() / "manifests" / "manifest_final.parquet")
TAGS = sorted({t for _, _, t in ARMS} | {"f320"})
have = crop.available_tags()
if [t for t in TAGS if t not in have]:
    raise SystemExit(f"[X] 태그 없음: {[t for t in TAGS if t not in have]}")
keep = crop.chunks_with_crops(df, TAGS)
df = df[df["chunk"].isin(keep)].reset_index(drop=True)
print(f"쓸 청크 {keep} — {len(df):,}행")

# ⚠️ 분할이 깨져 있으면 여기서 멈춥니다 (누수 확인)
split.verify(df, fold=0, strict=True)

s1 = stages.to_stage1(crop.switch_tag(df, "f320", verbose=False), verbose=False)
ho = split.get_holdout(s1).reset_index(drop=True)     # 🚨 정상 포함 holdout
y1 = (ho["label"] != stages.NORMAL_LABEL).to_numpy().astype(int)
print(f"\n🚨 holdout {len(ho):,}장  (정상 {int((y1 == 0).sum()):,} / "
      f"병변 {int(y1.sum()):,})")
print(f"   val 은 {len(split.get_fold(s1, 0)[1]):,}장이었습니다 (비교용)")
```

## 3. 1단계 → 넘긴 사진 고르기

```python
def temperature(exp):
    f = train.ckpt_dir(exp) / "temperature.json"
    return (float(json.loads(f.read_text(encoding="utf-8"))["temperature"])
            if f.exists() else 1.0)


def softmax(a):
    e = np.exp(a - a.max(1, keepdims=True))
    return e / e.sum(1, keepdims=True)


def infer(exp, frame, classes):
    m = models.build(train.model_key_from_exp(exp), n_classes=len(classes),
                     pretrained=False)
    sd = torch.load(train.ckpt_dir(exp) / "best.pt", map_location="cpu",
                    weights_only=False)
    m.load_state_dict(sd.get("model", sd.get("state_dict", sd)), strict=False)
    m = m.to(DEV).eval()
    cfg = CFG(); cfg.img_size = 384
    dl, ds = data.eval_loader(frame, cfg, model=m, classes=classes)
    if len(ds) != len(frame):
        raise SystemExit(f"[X] {exp}: {len(frame) - len(ds):,}행이 빠졌습니다")
    out = []
    with torch.no_grad():
        for x, _ in dl:
            out.append(m(x.to(DEV)).float().cpu())
    del m; torch.cuda.empty_cache()
    return torch.cat(out).numpy()


t0 = time.time()
T1 = temperature(STAGE1)
p1 = softmax(infer(STAGE1, ho, stages.CLASSES_STAGE1).astype(np.float64) / T1)[:, 1]
flag = p1 >= THR
n_fa = int((flag & (y1 == 0)).sum())
rec_ho = float(flag[y1 == 1].mean())
fa_ho = float(flag[y1 == 0].mean())
print(f"[1단계] T={T1:.4f} ({time.time()-t0:.0f}s)")
print(f"  넘긴 사진 {int(flag.sum()):,}장 — 헛알림 {n_fa:,}장 "
      f"({n_fa / max(int(flag.sum()), 1):.1%})")
print(f"  recall {rec_ho:.4f}   헛알림률 {fa_ho:.1%}")
print(f"  ↔ val: recall 0.9434 · 헛알림률 24.5%   "
      f"(차이 {rec_ho-0.9434:+.4f} / {fa_ho-0.245:+.1%})")
```

## 4. 2단계 — 넘긴 사진 전부에 (헛알림 포함)

```python
sub = ho[flag].reset_index(drop=True)
P2, names = {}, []
for name, exp, tag in ARMS:
    frame = crop.switch_tag(sub, tag, verbose=False)
    P2[name] = softmax(infer(exp, frame, CLASSES).astype(np.float64) / temperature(exp))
    names.append(name)
    print(f"[2단계] {name:16} 완료 ({time.time()-t0:.0f}s)")

truth = sub["label_orig"].to_numpy()
out = Path("/kaggle/working/step33_arrays_holdout.npz")
np.savez_compressed(out, p1=p1[flag], truth=truth,
                    **{f"p2_{i}": P2[n] for i, n in enumerate(names)},
                    arm_names=np.array(names, dtype=object))
print(f"\n★ 저장: {out}  ({out.stat().st_size/1e6:.1f}MB · {len(truth):,}행)")
```

## 5. 판정 — **기준은 `src/` 에 이미 박혀 있습니다**

여기서 기준을 새로 쓰지 않습니다. val 값과 나란히 찍어 **얼마나 떨어졌는지**만 봅니다.


```python
from src import experiments
from src.config import MORPH_GROUP, MORPH_GROUP_KEEP_A6, URGENCY_TIER

ens = np.mean([P2[n] for n in names], 0)
rel = P2[names[0]]
is_norm = truth == stages.NORMAL_LABEL
p1s = p1[flag]
tier_true = np.array([URGENCY_TIER.get(t, -1) for t in truth])
a6 = CLASSES.index("A6")
n = len(truth)

print(f"🚨 holdout · 1단계가 넘긴 사진 {n:,}장 (헛알림 {is_norm.mean():.1%} 포함)\n")
said = np.array(CLASSES, dtype=object)[ens.argmax(1)]
res = {"6종 이름": experiments.naming_report(
    {"conf": p1s * ens.max(1), "wrong": is_norm | (said != truth),
     "is_a6": truth == "A6", "said_a6": ens.argmax(1) == a6})}

print()
for label, mp in (("계열 4군 (A6 따로)", MORPH_GROUP_KEEP_A6),
                  ("계열 3군", MORPH_GROUP)):
    keys = list(dict.fromkeys(mp[c] for c in CLASSES))
    g = np.stack([ens[:, [i for i, c in enumerate(CLASSES) if mp[c] == k]].sum(1)
                  for k in keys], 1)
    gi = g.argmax(1)
    sd = np.array(keys, dtype=object)[gi]
    tr = np.array([mp.get(t, "정상") for t in truth], dtype=object)
    ts = np.array([max(URGENCY_TIER[c] for c in CLASSES if mp[c] == k)
                   for k in keys])[gi]
    res[label] = experiments.granularity_report(label, {
        "conf": p1s * g.max(1), "wrong": is_norm | (sd != tr),
        "tier_true": tier_true, "tier_said": ts})

# 릴리스 단독도 — 앙상블이 holdout 에서도 값을 하나
print("\n릴리스 단독 (앙상블이 holdout 에서도 값을 하나)")
keys = list(dict.fromkeys(MORPH_GROUP_KEEP_A6[c] for c in CLASSES))
g = np.stack([rel[:, [i for i, c in enumerate(CLASSES)
                      if MORPH_GROUP_KEEP_A6[c] == k]].sum(1) for k in keys], 1)
gi = g.argmax(1)
sd = np.array(keys, dtype=object)[gi]
tr = np.array([MORPH_GROUP_KEEP_A6.get(t, "정상") for t in truth], dtype=object)
ts = np.array([max(URGENCY_TIER[c] for c in CLASSES
                   if MORPH_GROUP_KEEP_A6[c] == k) for k in keys])[gi]
res["계열 4군 · 릴리스 단독"] = experiments.granularity_report(
    "계열 4군 · 릴리스 단독", {"conf": p1s * g.max(1), "wrong": is_norm | (sd != tr),
                        "tier_true": tier_true, "tier_said": ts})

# ── val 과 나란히 ──────────────────────────────────────────────
print("\n" + "=" * 68)
print("val(STEP 31·32) → holdout — **떨어지는 게 정상입니다**")
print("=" * 68)
VAL = {"6종 이름": 0.414, "계열 4군 (A6 따로)": 0.698, "계열 3군": 0.708,
       "계열 4군 · 릴리스 단독": 0.594}
print(f"  {'':24}{'val':>9}{'holdout':>10}{'차이':>9}")
for k, v in VAL.items():
    now = res[k]["coverage"]
    print(f"  {k:24}{v:>9.1%}{now:>10.1%}{now - v:>+9.1%}")

j = Path("/kaggle/working/step33_holdout.json")
j.write_text(json.dumps(
    {"step": "STEP 33 — holdout 확인", "chunks": keep,
     "n_holdout": int(len(ho)), "n_flagged": int(flag.sum()),
     "n_false_alarm": n_fa, "stage1_threshold": THR,
     "stage1_recall": rec_ho, "stage1_false_alarm_rate": fa_ho,
     "results": res, "val_for_comparison": VAL},
    indent=2, ensure_ascii=False, default=float), encoding="utf-8")
print(f"\n저장: {j}")
print("\n⚠️ 결과가 나빠도 **설정을 바꿔 다시 돌리지 마세요.** 그 순간")
print("   holdout 은 두 번째 val 이 되고 우리는 기댈 곳을 잃습니다.")
```
